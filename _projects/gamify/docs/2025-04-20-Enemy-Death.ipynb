  <h2>Modeling course data with a class</h2>
  <p class="lead">OCS has four courses — CSSE, CSP, CSA, and CSH. Each needs the same shape of data: a code, title, mission, unlock threshold, and a learning journey. A <em>class</em> lets us define that shape once and stamp out instances for each course.</p>

  <div class="concepts">
    <div class="concept"><div class="concept-label">Pattern</div><div class="concept-val">Class + new</div><div class="concept-sub">blueprint + instances</div></div>
    <div class="concept"><div class="concept-label">Courses</div><div class="concept-val">4 instances</div><div class="concept-sub">CSSE · CSP · CSA · CSH</div></div>
    <div class="concept"><div class="concept-label">Unlock gate</div><div class="concept-val">totalClicks</div><div class="concept-sub">threshold per course</div></div>
    <div class="concept"><div class="concept-label">Registry</div><div class="concept-val">window.ocsCourses</div><div class="concept-sub">global course lookup</div></div>
  </div>

  <p>Compare the two approaches below. Plain objects work, but a class gives you <em>shared behavior</em> and a single place to change the structure.</p>

  <div class="anno-wrap">
    <div class="anno-code">
      <div class="code-header"><span class="code-title">plain object vs class instance</span><span class="code-tag">JS</span></div>
      <div class="code-body"><pre><span class="cmt">// ❌ Plain object — data repeated, no shared logic</span>
<span class="kw">const</span> csseCourse = {
  code: <span class="str">'CSSE'</span>, title: <span class="str">'Software Engineering'</span>,
  unlockThreshold: <span class="num">5</span>,
  isUnlocked: (clicks) => clicks >= <span class="num">5</span>, <span class="cmt">// A — copy-pasted per course</span>
};

<span class="cmt">// ✅ Class — shape defined once, stamped out 4 times</span>
<span class="kw">class</span> <span class="fn">Course</span> {
  <span class="fn">constructor</span>(code, title, ...) {     <span class="cmt">// B — shared blueprint</span>
    <span class="kw">this</span>.<span class="prop">code</span> = code;
    <span class="kw">this</span>.<span class="prop">unlockThreshold</span> = unlockThreshold;
  }
  <span class="fn">isUnlocked</span>(totalClicks) {           <span class="cmt">// C — method defined once</span>
    <span class="kw">return</span> totalClicks >= <span class="kw">this</span>.<span class="prop">unlockThreshold</span>;
  }
}</pre></div>
    </div>
    <div class="anno-notes">
      <div class="anno-note"><div class="anno-icon">A</div><div class="anno-text"><strong>Logic duplication risk</strong> — with plain objects, every course would need its own copy of <code>isUnlocked</code>. If the rule changes, you'd edit 4 places instead of 1.</div></div>
      <div class="anno-note"><div class="anno-icon">B</div><div class="anno-text"><strong>constructor</strong> runs once when you write <code>new Course(...)</code>. It receives all the arguments and stores them as <code>this.property</code> — the instance's own data.</div></div>
      <div class="anno-note"><div class="anno-icon">C</div><div class="anno-text"><strong>Prototype method</strong> — <code>isUnlocked</code> lives on the <em>prototype</em>, not on each instance. All 4 course instances share the same function in memory.</div></div>
    </div>
  </div>

  <h2>How the constructor builds each course</h2>
  <p class="lead">The constructor receives 7 parameters. Most are required; <em>learningJourney</em> and <em>zoneTitle</em> have <em>defaults</em> so callers don't have to supply them.</p>

  <div class="anno-wrap">
    <div class="anno-code">
      <div class="code-header"><span class="code-title">Course constructor — full signature</span><span class="code-tag">JS</span></div>
      <div class="code-body"><pre><span class="kw">class</span> <span class="fn">Course</span> {
  <span class="fn">constructor</span>(
    code,                          <span class="cmt">// A — short ID e.g. 'CSSE'</span>
    title,                         <span class="cmt">// e.g. 'CSSE - Software Engineering'</span>
    subtitle,                      <span class="cmt">// e.g. 'Games, Foundation'</span>
    mission,                       <span class="cmt">// long description string</span>
    unlockThreshold,               <span class="cmt">// B — min clicks to unlock</span>
    learningJourney = [],          <span class="cmt">// C — default empty array</span>
    zoneTitle = <span class="str">''</span>                <span class="cmt">// D — default empty string</span>
  ) {
    <span class="kw">this</span>.<span class="prop">code</span>             = code;
    <span class="kw">this</span>.<span class="prop">title</span>            = title;
    <span class="kw">this</span>.<span class="prop">subtitle</span>         = subtitle;
    <span class="kw">this</span>.<span class="prop">mission</span>          = mission;
    <span class="kw">this</span>.<span class="prop">unlockThreshold</span>  = unlockThreshold;
    <span class="kw">this</span>.<span class="prop">learningJourney</span>  = learningJourney;
    <span class="kw">this</span>.<span class="prop">zoneTitle</span> = zoneTitle || <span class="cmt">// E — fallback expression</span>
      <span class="str">`${</span>code<span class="str">} Zone - ${</span>title<span class="str">}`</span>;
  }
}</pre></div>
    </div>
    <div class="anno-notes">
      <div class="anno-note"><div class="anno-icon">A</div><div class="anno-text"><strong>code</strong> is the short identifier used as a lookup key in <code>window.ocsCourses</code>. It doubles as the dictionary key: <code>ocsCourses['CSSE']</code>.</div></div>
      <div class="anno-note"><div class="anno-icon">B</div><div class="anno-text"><strong>unlockThreshold</strong> — courses get harder to unlock: CSSE=5, CSP=10, CSA=15, CSH=20. This integer is what both <code>isUnlocked()</code> and <code>getRemainingClicks()</code> use.</div></div>
      <div class="anno-note"><div class="anno-icon">C</div><div class="anno-text"><strong>Default parameter <code>= []</code></strong> — if the caller omits <code>learningJourney</code>, it gets an empty array, not <code>undefined</code>. Safe to call <code>.length</code> on immediately.</div></div>
      <div class="anno-note"><div class="anno-icon">D</div><div class="anno-text"><strong>Default parameter <code>= ''</code></strong> — same pattern for zoneTitle. An empty string is falsy in JS, so the next line's <code>||</code> expression kicks in when nothing is passed.</div></div>
      <div class="anno-note"><div class="anno-icon">E</div><div class="anno-text"><strong>Fallback expression</strong> — <code>zoneTitle || \`${code} Zone - ${title}\`</code> uses the short-circuit: if <code>zoneTitle</code> is falsy (empty string), the template literal is used instead. Callers can override it, but rarely need to.</div></div>

  <h2>Two methods, one job: the unlock gate</h2>
  <p class="lead">Both methods take <code>totalClicks</code> — the running click count from elsewhere in the game — and answer different questions about whether a course is accessible.</p>

  <div class="anno-wrap">
    <div class="anno-code">
      <div class="code-header"><span class="code-title">Course · isUnlocked() and getRemainingClicks()</span><span class="code-tag">JS</span></div>
      <div class="code-body"><pre>  <span class="fn">isUnlocked</span>(totalClicks) {
    <span class="kw">return</span> totalClicks >= <span class="kw">this</span>.<span class="prop">unlockThreshold</span>; <span class="cmt">// A</span>
  }

  <span class="fn">getRemainingClicks</span>(totalClicks) {
    <span class="kw">return</span> Math.<span class="fn">max</span>(                          <span class="cmt">// B</span>
      <span class="num">0</span>,
      <span class="kw">this</span>.<span class="prop">unlockThreshold</span> - totalClicks
    );
  }</pre></div>
    </div>
    <div class="anno-notes">
      <div class="anno-note"><div class="anno-icon">A</div><div class="anno-text"><strong>isUnlocked()</strong> returns a boolean. <code>>=</code> means "at threshold or past it" — exactly at 5 clicks, CSSE opens. The caller uses this for conditional rendering.</div></div>
      <div class="anno-note"><div class="anno-icon">B</div><div class="anno-text"><strong>Math.max(0, ...)</strong> clamps the result so it never goes negative. Without <code>Math.max</code>, a player with 25 clicks asking CSSE "how many more?" would get <code>5-25 = -20</code> — meaningless and potentially confusing.</div></div>
    </div>
  </div>

  <div class="flow">
